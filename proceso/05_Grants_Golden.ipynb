# 05 Grants - Permisos basicos
Consulta en Nintendo, Play y PC. Consulta y escritura en Xbox. Visibilidad del catalogo para Comunidad.

```sql
CREATE WIDGET TEXT aplicar_grants DEFAULT 'false';
CREATE WIDGET TEXT catalogo DEFAULT 'fi_videojuegos';
```

```sql
BEGIN
  SELECT assert_true(lower(:aplicar_grants) IN ('true', 'false'), 'aplicar_grants invalido');

  IF lower(:aplicar_grants) = 'true' THEN
    SELECT assert_true(:catalogo = 'fi_videojuegos', 'Estos permisos corresponden al catalogo fi_videojuegos');

    GRANT USE CATALOG ON CATALOG fi_videojuegos TO `Nintendo`;
    GRANT USE SCHEMA ON SCHEMA fi_videojuegos.bronze TO `Nintendo`;
    GRANT SELECT ON TABLE fi_videojuegos.bronze.fi_br_nintendo TO `Nintendo`;
    GRANT USE CATALOG ON CATALOG fi_videojuegos TO `videojuegos1@outlook.com`;
    GRANT USE SCHEMA ON SCHEMA fi_videojuegos.bronze TO `videojuegos1@outlook.com`;
    GRANT SELECT ON TABLE fi_videojuegos.bronze.fi_br_nintendo TO `videojuegos1@outlook.com`;
    GRANT USE CATALOG ON CATALOG fi_videojuegos TO `Play`;
    GRANT USE SCHEMA ON SCHEMA fi_videojuegos.bronze TO `Play`;
    GRANT SELECT ON TABLE fi_videojuegos.bronze.fi_br_playstation TO `Play`;
    GRANT USE CATALOG ON CATALOG fi_videojuegos TO `videojuegos2@outlook.com`;
    GRANT USE SCHEMA ON SCHEMA fi_videojuegos.bronze TO `videojuegos2@outlook.com`;
    GRANT SELECT ON TABLE fi_videojuegos.bronze.fi_br_playstation TO `videojuegos2@outlook.com`;
    GRANT USE CATALOG ON CATALOG fi_videojuegos TO `Xbox`;
    GRANT USE SCHEMA ON SCHEMA fi_videojuegos.bronze TO `Xbox`;
    GRANT SELECT, MODIFY ON TABLE fi_videojuegos.bronze.fi_br_xbox TO `Xbox`;
    GRANT USE CATALOG ON CATALOG fi_videojuegos TO `videojuego3@outlook.com`;
    GRANT USE SCHEMA ON SCHEMA fi_videojuegos.bronze TO `videojuego3@outlook.com`;
    GRANT SELECT, MODIFY ON TABLE fi_videojuegos.bronze.fi_br_xbox TO `videojuego3@outlook.com`;
    GRANT USE CATALOG ON CATALOG fi_videojuegos TO `PC`;
    GRANT USE SCHEMA ON SCHEMA fi_videojuegos.golden TO `PC`;
    GRANT SELECT ON TABLE fi_videojuegos.golden.fi_gd_ranking_familia_anio TO `PC`;
    GRANT USE CATALOG ON CATALOG fi_videojuegos TO `videojuego4@outlook.com`;
    GRANT USE SCHEMA ON SCHEMA fi_videojuegos.golden TO `videojuego4@outlook.com`;
    GRANT SELECT ON TABLE fi_videojuegos.golden.fi_gd_ranking_familia_anio TO `videojuego4@outlook.com`;
    GRANT BROWSE ON CATALOG fi_videojuegos TO `Comunidad`;
    GRANT BROWSE ON CATALOG fi_videojuegos TO `videojuego5@outlook.com`;

    SELECT 'Permisos basicos aplicados' AS estado;
  ELSE
    SELECT 'Permisos omitidos. Usa aplicar_grants=true para aplicarlos.' AS estado;
  END IF;
END;
```
